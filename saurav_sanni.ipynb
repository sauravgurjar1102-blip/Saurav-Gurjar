```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/adult saurav.csv')
df
```

```python
age_array = df["age"].to_numpy()
hours_array = df["hours.per.week"].to_numpy()

print("Age NumPy Array:")
print(age_array[:10])

print("\nHours per Week NumPy Array:")
print(hours_array[:10])
```

```python
print("First age:", age_array[0])
print("Fifth age:", age_array[4])
print("Last age:", age_array[-1])
```

```python
# Basic Slicing

print("First 10 values of age array:")
print(age_array[:10])

print("\nBasic Slice (index 2 to 6):")
print(age_array[2:7])
```

```python
age_array[2:7]
```

```python
age_25_35 = age_array[(age_array >= 25) & (age_array <= 35)]

print("Ages between 25 and 35:")
print(age_25_35[:20])
```

```python
# Basic Slicing vs Advanced (Fancy) Indexing

print("Original Array - First 10 values:")
print(age_array[:10])

# Basic slicing
basic = age_array[0:5]

# Fancy indexing
fancy = age_array[[0, 2, 4]]

print("\nBasic Slicing [0:5]:")
print(basic)

print("\nAdvanced/Fancy Indexing [[0, 2, 4]]:")
print(fancy)
```

```python
age_array[0:5]
```

```python
age_array[[0,2,4]]
```

```python
# Boolean Indexing

condition = (age_array >= 25) & (age_array <= 35)

print("Boolean Condition:")
print(condition[:20])

print("\nAges between 25 and 35:")
print(age_array[condition][:20])
```
